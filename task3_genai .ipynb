```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv()
```

```python
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
```

```python
openrouter_client = OpenAI(
    api_key = OPENROUTER_API_KEY,
    base_url = "https://openrouter.ai/api/v1"
)
```

```python
messages = [{"role":"system", "content":"you are a conversational botmy name"}]
print("Type START to begin chatting with AI bot, type STOP to end the chat.")
while True:
    user_input = input("Enter your text here: ")
    messages.append(
        {"role":"user", "content":user_input}
    )
    if user_input == "STOP":
        print("Your chat has been ended")
        break
    response = openrouter_client.responses.create(
        model = "inclusionai/ling-3.0-flash-fin:free",
        input = messages,
        tools = [
            {"type":"web_search"}
        ]
    
    )
    messages.append(
        {"role":"assistant", "content":response.output_text}
    )
    print(response.output_text)
```
