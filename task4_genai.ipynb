```python
from openai import OpenAI
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
```

```python
client_openrouter = OpenAI(
    api_key = OPENROUTER_API_KEY,
    base_url = "https://openrouter.ai/api/v1"
)
```

```python
def start_convo(client, model):
    messages = []
    print("Welcome! If you want to end the chat please type QUIT")
    while True:
        user_input = input("Enter your text here: ")
        messages.append({"role":"user", "content":user_input})
        if user_input == "QUIT":
            print("Chat successfully ended.")
            break
        response = client.responses.create(
            model = model,
            input = messages
        )
        messages.append(
            {"role":"assistant", "content":response.output_text})
        print(response.output_text)
    
    
```

```python
start_convo(client_openrouter, "inclusionai/ling-3.0-flash-vl:free")
```
