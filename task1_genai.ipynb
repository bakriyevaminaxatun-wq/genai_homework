**Math Test Practice Bot: An interactive tutoring chatbot that quizzes users, rates answers, and explains mistakes.**

```python
from openai import OpenAI
import os
from dotenv import load_dotenv
```

```python
os.getcwd()
```

```python
load_dotenv()
```

```python
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
```

```python
sys_message = """
you are a chatbot for practicing for a math test. give questions to the user, 
check the answer, and give them a score out of 10 based on their explanation.
if a user made any mistakes, explain it step by step. then give another question till the user says stop.
"""
messages = [{"role":"system", "content":sys_message}]
```

```python
openrouter_client = OpenAI(
    api_key = OPENROUTER_API_KEY,
    base_url = "https://openrouter.ai/api/v1"
)
```

```python
print("This is a chatbot to practice for your math exam. Type START to get questions. If you feel ready for the test, type STOP to end the conversation")
while True:
    user_input = input("Insert text: ")
    messages.append(
        {"role":"user", "content":user_input}
    )
    if user_input == "STOP":
        print("Your chat ended. Good luck on your math test!")
        break
    response = openrouter_client.responses.create(
    model = "inclusionai/ling-3.0-flash-sante:free",
    input = messages)
    messages.append(
        {"role":"assistant", "content": response.output_text}
    )    
    print(response.output_text)
    
```
