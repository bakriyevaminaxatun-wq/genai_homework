```python
from openai import OpenAI
from dotenv import load_dotenv
import os
```

```python
os.getcwd()
```

```python
load_dotenv()
```

```python
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
```

```python
sys_message = """
you are a python tutor. if questions are not related to Python, say - I cant answer this.
if questions is in another language, say you can only talk in eng
explain questions clearly and give tasks for topic explanation.
"""
messages = [{"role":"system","content":sys_message}]
```

```python
openrouter_client = OpenAI(
    api_key = OPENROUTER_API_KEY,
    base_url = "https://openrouter.ai/api/v1"
)
```

```python
print("Ask any question about Python programming language")
while True:
    user_input = input("Enter your text here: ")
    messages.append({"role":"user", "content":user_input})
    if user_input == "STOP":
        print("chat ended")
        break
    response = openrouter_client.responses.create(
        model = "inclusionai/ling-3.0-flash-fin:free",
        input = messages,
        max_output_tokens = 250
    )
    messages.append({"role": "assistant", "content": response.output_text})
    print(response.output_text) 
```

```python
messages.clear()
```
